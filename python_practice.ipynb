## Understanding `.strip()`

The `.strip()` method is used to remove leading and trailing whitespace (spaces, tabs, newlines) from a string. You can also specify characters to remove from the beginning and end of the string.

```python
# Example 1: Removing whitespace
text_with_spaces = "   Hello, World!   \n"
cleaned_text = text_with_spaces.strip()
print(f"Original: '{text_with_spaces}'")
print(f"Cleaned: '{cleaned_text}'")

# Example 2: Removing specific characters
data_string = "###DataValue###"
cleaned_data = data_string.strip('#')
print(f"Original: '{data_string}'")
print(f"Cleaned: '{cleaned_data}'")
```

## Understanding `.split()`

The `.split()` method is used to split a string into a list of substrings. By default, it splits the string by any whitespace character and removes empty strings from the result. You can also specify a custom delimiter to split the string by.

```python
# Example 1: Splitting by whitespace (default)
sentence = "This is a sample sentence."
words = sentence.split()
print(f"Original sentence: '{sentence}'")
print(f"Split into words: {words}")

# Example 2: Splitting by a comma
csv_data = "apple,banana,orange,grape"
fruits = csv_data.split(',')
print(f"Original CSV data: '{csv_data}'")
print(f"Split into fruits: {fruits}")

# Example 3: Splitting by a specific character with max splits
path = "/usr/local/bin/python"
parts = path.split('/', 2) # Split at most 2 times
print(f"Original path: '{path}'")
print(f"Split path parts (max 2): {parts}")
```

```python
#nested if
# --- Movie Ticket Pricing System ---
BASE_PRICE = 50.0

# 1. Collect Inputs
print("--- Welcome to the Cinema ---")
customer_age = int(input("Enter your age: "))
showtime_hour = int(input("Enter the showtime hour (0-23): "))
promo_code = input("Enter promo code (or press Enter to skip): ")

# 2. Determine Base Discount based on Age
if customer_age < 12:
    price = BASE_PRICE * 0.50
    discount_msg = "50% Child Discount Applied!"

elif customer_age >= 60:
    price = BASE_PRICE * 0.70
    discount_msg = "30% Senior Discount Applied!"

elif 12 <= customer_age < 60:
    # Check for Matinee (Before 6:00 PM / 18:00)
    if showtime_hour < 18:
        price = BASE_PRICE * 0.90
        discount_msg = "10% Matinee Discount Applied!"
    else:
        price = BASE_PRICE
        discount_msg = "Full Price (No Matinee Discount)"

# 3. Apply Extra Promo Code (Optional)
# .upper() makes "save5" or "Save5" work just like "SAVE5"
if promo_code.upper() == "SAVE5":
    price = price - 5.0
    discount_msg += " + SAVE5 Promo (-$5.00)"

# 4. Final Output
print("-" * 30)
print(f"Status: {discount_msg}")
print(f"Your Final Ticket Price is: ${price:,.2f}")
print("-" * 30)
```

```python
annual_income=float(input("Enter your annual income: "))
if annual_income >= 100000:
  tax=annual_income*0.2
  print(f"Your tax is ${tax:,.2f}")
elif annual_income >= 50000 and annual_income < 100000:
  tax=annual_income*0.1
  print(f"Your tax is ${tax:,.2f}")
else:
  tax=annual_income*0.05
  print(f"Your tax is ${tax:,.2f}")
```
