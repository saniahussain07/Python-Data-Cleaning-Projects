## Understanding `.strip()`

The `.strip()` method is used to remove leading and trailing whitespace (spaces, tabs, newlines) from a string. You can also specify characters to remove from the beginning and end of the string.

```python
# Example 1: Removing whitespace
text_with_spaces = "   Hello, World!   \n"
cleaned_text = text_with_spaces.strip()
print(f"Original: '{text_with_spaces}'")
print(f"Cleaned: '{cleaned_text}'")

# Example 2: Removing specific characters
data_string = "###DataValue###"
cleaned_data = data_string.strip('#')
print(f"Original: '{data_string}'")
print(f"Cleaned: '{cleaned_data}'")
```

## Understanding `.split()`

The `.split()` method is used to split a string into a list of substrings. By default, it splits the string by any whitespace character and removes empty strings from the result. You can also specify a custom delimiter to split the string by.

```python
# Example 1: Splitting by whitespace (default)
sentence = "This is a sample sentence."
words = sentence.split()
print(f"Original sentence: '{sentence}'")
print(f"Split into words: {words}")

# Example 2: Splitting by a comma
csv_data = "apple,banana,orange,grape"
fruits = csv_data.split(',')
print(f"Original CSV data: '{csv_data}'")
print(f"Split into fruits: {fruits}")

# Example 3: Splitting by a specific character with max splits
path = "/usr/local/bin/python"
parts = path.split('/', 2) # Split at most 2 times
print(f"Original path: '{path}'")
print(f"Split path parts (max 2): {parts}")
```

```python
def calculate_marketing_roi(investment, revenue):
  if investment <= 0:
    return "Invalid investment amount.",0,0 # Return 3 values consistently
  roi = ((revenue - investment) / investment) * 100
  net_profit = revenue - investment
  if roi >= 50:
    return "High Performing Campaign.",roi,  net_profit
  elif 10 <= roi < 50:
    return "Stable.",roi,  net_profit
  else:
    return "Underperforming - Needs Review.",roi,  net_profit

# Fix: Apply .replace() to the string input BEFORE converting to float
user_inv_str = input("Enter the investment amount: ")
user_inv = float(user_inv_str.replace(",", ""))

user_rev_str = input("Enter the revenue amount: ")
user_rev = float(user_rev_str.replace(",", ""))

status, roi, net_profit = calculate_marketing_roi(user_inv,user_rev)
print("-"*25)
print(f"ROI Percentage: {roi:,.2f}%") # Added % sign for clarity
print(f"Analysis: {status}")
print(f"Net Profit: $ {net_profit:,.2f}") # Added $ sign for clarity
print("-"*25)


import seaborn as sns
import matplotlib.pyplot as plt

# --- 1. GET THE DATA (Automation starts here) ---
user_inv = float(input("Enter Investment: ").replace(",", ""))
user_rev = float(input("Enter Revenue: ").replace(",", ""))

# Calculate ROI and Profit (Using your logic)
roi = ((user_rev - user_inv) / user_inv) * 100
profit = user_rev - user_inv

# --- 2. UPDATE THE VISUAL ---
sns.set_theme(style="whitegrid")
# This dictionary NOW uses the new numbers you just typed
data = {'Category': ['Investment', 'Profit'], 'Amount': [user_inv, profit]}

plt.figure(figsize=(8, 5))
ax = sns.barplot(x='Category', y='Amount', data=data, palette=['#FF9999', '#66B3FF'], hue='Category', legend=False)

# Add dynamic title and labels
plt.title(f'Business Growth Insight (ROI: {roi:.2f}%)', fontsize=15)
for p in ax.patches:
    ax.annotate(f'{p.get_height():,.0f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 9), textcoords='offset points', fontweight='bold')

plt.show()
```

Challenge 2: The Multi-Currency Salary Converter 🌍
Goal: Analysts often deal with global data. You need a function that handles different currencies.

Task: Write a function convert_salary(amount, currency).

Logic:

Use a dictionary inside the function for exchange rates (e.g., USD: 278, CNY: 38, AED: 75).

The function should take a PKR amount and return the value in the chosen currency.

Bonus: Add a "Tax" parameter that subtracts 10% before converting.

```python
def convert_salary(amount_pkr, currency, tax=0.10):
  exchange_rates = {
      "USD":279.20,
      "CNY":40.56,
      "AED":76.1,
  }
  symbols = {
        "USD": "$",
        "CNY": "¥",
        "AED": "AED "
    }

  amount_after_tax = amount_pkr * (1 - tax)
  if currency in exchange_rates:
    converted_amount = amount_after_tax / exchange_rates[currency]
    return converted_amount, symbols[currency]
  else:
    # Consistently return two values, even if currency is not supported
    return None, "currency not supported"

salary_pkr_str = input("Enter your salary in PKR:")
salary_pkr = float(salary_pkr_str.replace(",", ""))
target_curr = input("Enter the currency to convert to(USD, CNY, AED): ").upper()

final_salary, sym = convert_salary(salary_pkr, target_curr)

# Check if final_salary is not None to determine if conversion was successful
if final_salary is not None:
 print("-"*30)
 print(f"Your salary in {target_curr} is {sym}{final_salary:,.2f}")
 print("-"*30)
else:
 print("-"*30)
 print(f"Error: {sym} for {target_curr}") # sym now contains the error message
 print("-"*30)
```

Challenge 4: The Cricket Strike Rate Master 🏏Goal: Advanced logic using functions.Task: Write a function analyze_batsman(runs, balls).Formula: $Strike Rate = \frac{Runs}{Balls} \times 100$Logic:If the Strike Rate is > 150, categorize as "Aggressive."If between 100-150, "Consistent."Below 100, "Anchor."Edge Case: Ensure your function doesn't crash if balls is 0 (use an if statement to handle this!).

```python
def analyze_batsman(runs, balls):
  if balls == 0:
    return "Cannot calculate strike rate with zero balls.", 0  # Return error message and SR 0
  strike_rate = (runs / balls) * 100
  if strike_rate > 150:
    return "Aggressive", strike_rate
  elif 100 <= strike_rate <= 150:
    return "Consistent", strike_rate
  else:
    return "Anchor", strike_rate

runs = float(input("Enter the number of runs: "))
balls = float(input("Enter the number of balls: "))
result, sr = analyze_batsman(runs, balls)

# Check if the result is an error message or a category
if isinstance(result, str) and "Cannot calculate" in result:
  print(result)
else:
  print(f"The batsman's strike rate is {sr:.2f} and categorized as {result}.")
```

Challenge 1: The Descriptive Statistics SummarizerGoal: In data science, you rarely look at one number; you look at the "spread" of data. Create a function that provides a quick summary of any numerical list (like cricket scores or sales data).Task: Write a function get_stats_summary(data_list).Logic:Calculate the Mean (Average): $\bar{x} = \frac{\sum x}{n}$Find the Range (Difference between maximum and minimum values).Return: A dictionary containing both values.Example Input: [85, 90, 78, 92, 88]

```python
def get_stats_summary(data_list):
  mean = sum(data_list) / len(data_list)
  data_range = max(data_list) - min(data_list)
  return {"Mean": mean, "Range": data_range}

sales_data = [5000, 4500, 6500, 2000]
summary = get_stats_summary(sales_data)
print(summary)
```

Challenge 2: The Business "Break-Even" AnalyzerGoal: Businesses need to know the exact point where they stop losing money and start making a profit. This is a core analytical skill for commerce and business intelligence.Task: Write a function calculate_break_even(fixed_costs, price_per_unit, cost_per_unit).Formula: $Units = \frac{Fixed Costs}{Price - Variable Cost}$Logic: * The function should calculate how many units must be sold to cover all costs.Edge Case: If price_per_unit is less than or equal to cost_per_unit, return a warning: "Business model is not sustainable."

```python
def calculate_break_even(fixed_costs, price_per_unit, cost_per_unit):
  # If price per unit is less than or equal to cost per unit, the model is not sustainable.
  if price_per_unit <= cost_per_unit:
    if price_per_unit == cost_per_unit:
      return "The profit per unit is zero (price per unit should be > cost per unit).", None
    else:
      return "Business model is not sustainable (price per unit < cost per unit).", None
  else:
    # Calculate units needed to break even
    units = fixed_costs / (price_per_unit - cost_per_unit)
    return "Business model is sustainable.", units

fixed_costs = float(input("Enter the fixed costs: "))
price_per_unit = float(input("Enter the price per unit: "))
cost_per_unit = float(input("Enter the cost per unit: "))

msg, units = calculate_break_even(fixed_costs, price_per_unit, cost_per_unit)

print("----------------------------------")
print(f"Status: {msg}")
if units is not None:
  print(f"Break-even units: {units:,.2f}")
else:
  print("Cannot calculate break-even units due to unsustainable model.")
print("----------------------------------")
```

Challenge 3: HSK Character Practice Goal CheckerGoal: Track your progress in learning new languages by automating the evaluation of study sessions.Task: Create a function evaluate_hsk_session(chars_learned, minutes_spent).Logic:Calculate "Efficiency" as: $Efficiency = \frac{Characters}{Minutes}$If efficiency is > 0.5 (more than 1 character every 2 minutes), return "Highly Productive."If between 0.2 and 0.5, return "Steady Progress."Else, return "Focus on Quality over Quantity."

```python
def evaluate_hsk_session(chars_learned, minutes_spent):
  if minutes_spent == 0:
    return "Cannot calculate efficiency with zero minutes spent.", 0
  Efficiency = chars_learned / minutes_spent
  if Efficiency >= 0.5:
    return "Highly Productive", Efficiency
  elif 0.2 <= Efficiency < 0.5:
    return "Steady Progress", Efficiency
  else:
    return "Focus on Quality over Quantity", Efficiency

chars_learned = int(input("Enter the number of characters learned: "))
minutes_spent = int(input("Enter the number of minutes spent: "))
status, efficiency = evaluate_hsk_session(chars_learned, minutes_spent)
print(f"Efficiency: {efficiency:.2f}")
print(f"Message: {status}")
```

Challenge 4: Technical Skill - Password Strength Validator
Goal: This "Wildcard" task builds your logic for data cleaning and security—essential for any analyst handling sensitive databases.

Task: Write a function is_secure_password(password_string).

Requirements:

Length must be at least 8 characters.

Must contain at least one special character (e.g., @, #, $, !).

Return: True if it meets both, otherwise False.

Hint: Use a for loop inside your function to check each character in the strin

```python
def is_secure_password(password_string):
  special_characters = ['@', '#', '$', '!']
  if len(password_string) < 8:
    return False
  found_special = False
  for char in password_string:
    if char in special_characters:
      found_special = True
      break
  return found_special
password = input("Enter a password: ")
if is_secure_password(password):
    print("Password is secure.")
else:
    print("Password is not secure, Length must be at least 8 characters and must contain at least one special character (e.g., @, #, $,'!'.")
```

Challenge 5: Compound Growth PredictorGoal: Analysts use this to project future sales or the growth of a user base over time.Task: Write a function predict_growth(initial_value, rate, years).Formula: $Future Value = P(1 + r)^t$Logic: * P is the initial value, r is the growth rate (e.g., 0.05 for 5%), and t is the number of years.The function should return the final projected number rounded to 2 decimal places.

```python
def predict_growth(initial_value, rate, years):
   future_value = initial_value * (1 + rate) ** years
   return round(future_value, 2)

initial_value = input("Enter the initial value: ")
initial_value = float(initial_value.replace(",", ""))
rate = float(input("Enter the growth rate (e.g., 0.05 for 5%): "))
years = int(input("Enter the time period (e.g., 1,2,3):"))

future_value = predict_growth(initial_value, rate, years)
print(f"Future Value: {future_value}")
```
